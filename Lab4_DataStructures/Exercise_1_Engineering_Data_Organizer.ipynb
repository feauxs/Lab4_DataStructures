```python
# Engineering Test Data Organizer
LAST_NAME = "Campos"
SEED = 7
FAVORITE_ARTIST = "Car Seat Headrest"

def artist_positions(artist):
    positions = []
    for character in artist.lower():
        if character.isalpha():
            positions.append(ord(character) - ord("a") + 1)
    return positions

ARTIST_POSITIONS = artist_positions(FAVORITE_ARTIST)

def generated_number(index, offset=0, minimum=1, maximum=100):
    letter_value = ARTIST_POSITIONS[index % len(ARTIST_POSITIONS)]
    span = maximum - minimum + 1
    return minimum + (
        (SEED + letter_value + index * 3 + offset) % span
    )

# Display data in a formatted table
def print_table(headers, rows):
    rows = [[str(value) for value in row] for row in rows]
    headers = [str(header) for header in headers]

    widths = [
        max(len(headers[i]), max(
            (len(row[i]) for row in rows), default=0
        ))
        for i in range(len(headers))
    ]

    header_line = " | ".join(
        headers[i].ljust(widths[i])
        for i in range(len(headers))
    )
    separator = "-+-".join("-" * width for width in widths)

    print(header_line)
    print(separator)

    for row in rows:
        print(" | ".join(
            row[i].ljust(widths[i])
            for i in range(len(headers))
        ))

# 1. Generate engineering test records
sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["Temperature", "Voltage", "Current", "Pressure", "Speed"]
ranges = {
    "Temperature": (15, 100),
    "Voltage": (1, 240),
    "Current": (1, 30),
    "Pressure": (80, 200),
    "Speed": (10, 180)
}

records = []
index = 0

for source_index, source in enumerate(sources):
    selected_categories = (
        categories if source_index % 2 == 0
        else categories[:3] + categories[4:]
    )

    for category_index, category in enumerate(selected_categories):
        low, high = ranges[category]
        value = generated_number(
            index,
            offset=source_index + category_index,
            minimum=low,
            maximum=high
        )

        records.append({
            "source": source,
            "category": category,
            "value": value
        })
        index += 1

# Required repeated complete record
records.append(records[0].copy())

print("=== EXERCISE 1: ENGINEERING TEST DATA ORGANIZER ===")
print("\n1. GENERATED TEST DATA")

print_table(
    ["Source", "Category", "Value"],
    [[r["source"], r["category"], r["value"]] for r in records]
)

# 2. Organize records by source and category
organized = sorted(
    records,
    key=lambda r: (r["source"], r["category"], r["value"])
)

print("\n2. ORGANIZED TEST DATA")
print_table(
    ["Source", "Category", "Value"],
    [[r["source"], r["category"], r["value"]] for r in organized]
)

# 3. Detect repeated and distinct records
seen = set()
duplicates = []
for record in records:
    key = (record["source"], record["category"], record["value"])

    if key in seen:
        duplicates.append(record)
    else:
        seen.add(key)
print("\n3. REPEATED AND DISTINCT DATA")
print_table(
    ["Measurement", "Result"],
    [
        ["Total records", len(records)],
        ["Distinct complete records", len(seen)],
        ["Repeated complete records", len(duplicates)]
    ]
)

print("\nRepeated Records")
print_table(
    ["Source", "Category", "Value"],
    [[r["source"], r["category"], r["value"]] for r in duplicates]
)

print("\nDistinct Sources")
print(", ".join(sorted({r["source"] for r in records})))
print("\nDistinct Categories")
print(", ".join(sorted({r["category"] for r in records})))

# 4. Numerical summaries by category
print("\n4. NUMERICAL SUMMARIES BY CATEGORY")
numerical_summary = []
for category in categories:
    values = [
        r["value"] for r in records
        if r["category"] == category
    ]

    if values:
        numerical_summary.append([
            category,
            len(values),
            min(values),
            max(values),
            f"{sum(values) / len(values):.2f}",
            sum(values)
        ])

print_table(
    ["Category", "Count", "Minimum", "Maximum", "Mean", "Total"],
    numerical_summary
)

# 5. Categorical summaries
print("\n5. CATEGORICAL SUMMARIES BY SOURCE")
print_table(
    ["Source", "Number of Records"],
    [
        [source, sum(1 for r in records if r["source"] == source)]
        for source in sources
    ]
)

print("\nCATEGORICAL SUMMARIES BY CATEGORY")
print_table(
    ["Category", "Number of Records"],
    [
        [
            category,
            sum(1 for r in records if r["category"] == category)
        ]
        for category in categories
    ]
)

# 6. Ordered results
print("\n6. ORDERED RESULTS (by source, category, and value)")
print_table(
    ["Source", "Category", "Value"],
    [[r["source"], r["category"], r["value"]] for r in organized]
)

# 7. Final summary
print("\n7. FINAL SUMMARY")
print_table(
    ["Summary Item", "Result"],
    [
        ["Sources", len(sources)],
        ["Categories", len(categories)],
        ["Total measurement records", len(records)],
        ["Unique complete records", len(seen)],
        ["Repeated complete records", len(duplicates)]
    ]
)

print("\nGENERATION INPUTS")
print(f"Last name: {LAST_NAME}")
print(f"Seed: {SEED}")
print(f"Favorite artist: {FAVORITE_ARTIST}")
```
