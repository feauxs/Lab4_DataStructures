```python
# Student Performance Data Analyzer
LAST_NAME = "Campos"
SEED = 7
FAVORITE_ARTIST = "Car Seat Headrest"

def artist_positions(artist):
    positions = []
    for character in artist.lower():
        if character.isalpha():
            positions.append(ord(character) - ord("a") + 1)
    return positions

ARTIST_POSITIONS = artist_positions(FAVORITE_ARTIST)

def generated_number(index, offset=0, minimum=1, maximum=100):
    letter_value = ARTIST_POSITIONS[index % len(ARTIST_POSITIONS)]
    span = maximum - minimum + 1
    return minimum + (
        (SEED + letter_value + index * 3 + offset) % span
    )

# Display data in a formatted table
def print_table(headers, rows):
    rows = [[str(value) for value in row] for row in rows]
    headers = [str(header) for header in headers]
    widths = [
        max(
            len(headers[i]),
            max((len(row[i]) for row in rows), default=0)
        )
        for i in range(len(headers))
    ]

    print(" | ".join(
        headers[i].ljust(widths[i])
        for i in range(len(headers))
    ))
    print("-+-".join("-" * width for width in widths))

    for row in rows:
        print(" | ".join(
            row[i].ljust(widths[i])
            for i in range(len(headers))
        ))

# 1. Generate student performance records
students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["Programming", "Mathematics", "Electronics", "Physics", "English"]
records = []
index = 0

for student_index, student in enumerate(students):
    for subject_index, subject in enumerate(subjects):
        score = generated_number(
            index,
            offset=student_index + subject_index,
            minimum=50,
            maximum=100
        )

        records.append({
            "student": student,
            "subject": subject,
            "score": score
        })
        index += 1

# Required repeated student-subject record
records.append(records[0].copy())

# Required incomplete record (missing score)
records.append({
    "student": students[1],
    "subject": "Physics",
    "score": None
})

# Classify performance
def performance_category(score):
    if score is None:
        return "Incomplete"
    if score >= 90:
        return "Excellent"
    if score >= 80:
        return "Very Good"
    if score >= 70:
        return "Satisfactory"
    if score >= 60:
        return "Needs Improvement"
    return "Below Expectations"
for record in records:
    record["category"] = performance_category(record["score"])

print("=== EXERCISE 2: STUDENT PERFORMANCE DATA ANALYZER ===")

# 1. Generated performance records
print("\n1. GENERATED PERFORMANCE RECORDS")
print_table(
    ["Student", "Subject", "Score", "Category"],
    [
        [
            r["student"],
            r["subject"],
            "Missing" if r["score"] is None else r["score"],
            r["category"]
        ]
        for r in records
    ]
)

valid_records = [r for r in records if r["score"] is not None]
incomplete_records = [r for r in records if r["score"] is None]

# 2. Organized student-subject data
print("\n2. ORGANIZED STUDENT-SUBJECT DATA")
organized = sorted(
    records,
    key=lambda r: (
        r["student"],
        r["subject"],
        r["score"] is None
    )
)

print_table(
    ["Student", "Subject", "Score", "Category"],
    [
        [
            r["student"],
            r["subject"],
            "Missing" if r["score"] is None else r["score"],
            r["category"]
        ]
        for r in organized
    ]
)

# 3. Detect repeated student-subject pairs and incomplete records
pair_counts = {}

for r in records:
    pair = (r["student"], r["subject"])
    pair_counts[pair] = pair_counts.get(pair, 0) + 1
repeated_pairs = {
    pair: count
    for pair, count in pair_counts.items()
    if count > 1
}

print("\n3. REPEATED OR INCOMPLETE RECORDS")
print_table(
    ["Student", "Subject", "Record Count"],
    [
        [student, subject, count]
        for (student, subject), count in repeated_pairs.items()
    ]
)

print("\nIncomplete Records")
print_table(
    ["Student", "Subject", "Score", "Category"],
    [
        [r["student"], r["subject"], "Missing", r["category"]]
        for r in incomplete_records
    ]
)

print_table(
    ["Score Status", "Count"],
    [
        ["Valid scores", len(valid_records)],
        ["Missing scores", len(incomplete_records)]
    ]
)

# 4. Performance category summary
print("\n4. PERFORMANCE CATEGORIES")
category_names = [
    "Excellent",
    "Very Good",
    "Satisfactory",
    "Needs Improvement",
    "Below Expectations",
    "Incomplete"
]

category_counts = {
    category: sum(1 for r in records if r["category"] == category)
    for category in category_names
}

print_table(
    ["Performance Category", "Number of Records"],
    [[category, category_counts[category]] for category in category_names]
)

# 5. Student summaries
print("\n5. STUDENT PERFORMANCE SUMMARIES")
student_summary = []
for student in students:
    scores = [
        r["score"] for r in valid_records
        if r["student"] == student
    ]

    if scores:
        student_summary.append([
            student,
            len(scores),
            f"{sum(scores) / len(scores):.2f}",
            max(scores),
            min(scores)
        ])

print_table(
    ["Student", "Valid Scores", "Average", "Highest", "Lowest"],
    student_summary
)

# Subject summaries
print("\nSUBJECT PERFORMANCE SUMMARIES")
subject_summary = []
for subject in subjects:
    scores = [
        r["score"] for r in valid_records
        if r["subject"] == subject
    ]

    if scores:
        subject_summary.append([
            subject,
            len(scores),
            f"{sum(scores) / len(scores):.2f}",
            max(scores),
            min(scores)
        ])

print_table(
    ["Subject", "Valid Scores", "Average", "Highest", "Lowest"],
    subject_summary
)

# 6. Ordered results by subject, then student
print("\n6. ORDERED RESULTS")

ordered_results = sorted(
    records,
    key=lambda r: (r["subject"], r["student"])
)
print_table(
    ["Subject", "Student", "Score", "Category"],
    [
        [
            r["subject"],
            r["student"],
            "Missing" if r["score"] is None else r["score"],
            r["category"]
        ]
        for r in ordered_results
    ]
)

# 7. Final student performance report
print("\n7. FINAL STUDENT PERFORMANCE REPORT")
all_scores = [r["score"] for r in valid_records]
overall_average = (
    f"{sum(all_scores) / len(all_scores):.2f}"
    if all_scores else "N/A"
)

print_table(
    ["Summary Item", "Result"],
    [
        ["Students", len(students)],
        ["Subjects", len(subjects)],
        ["Total records", len(records)],
        ["Valid scores", len(valid_records)],
        ["Incomplete records", len(incomplete_records)],
        ["Repeated student-subject pairs", len(repeated_pairs)],
        ["Overall valid-score average", overall_average]
    ]
)

print("\nGENERATION INPUTS")
print(f"Last name: {LAST_NAME}")
print(f"Seed: {SEED}")
print(f"Favorite artist: {FAVORITE_ARTIST}")
print("\nPERFORMANCE BANDS")
print("90-100: Excellent")
print("80-89: Very Good")
print("70-79: Satisfactory")
print("60-69: Needs Improvement")
print("Below 60: Below Expectations")
```
